```python
import numpy as np
import pandas as pd
import scipy.stats as stats
from statsmodels.stats.multitest import multipletests

DEMO_FILE = "DEMO26Q1.txt"
DRUG_FILE = "DRUG26Q1.txt"
REAC_FILE = "REAC26Q1.txt"

demo = pd.read_csv(DEMO_FILE, sep="$", usecols=["primaryid", "sex"], low_memory=False)
drug = pd.read_csv(
    DRUG_FILE, sep="$", usecols=["primaryid", "drugname"], low_memory=False
)
reac = pd.read_csv(
    REAC_FILE, sep="$", usecols=["primaryid", "pt"], low_memory=False
)

tamsulosin_case = drug[
    drug["drugname"].str.contains("tamsulosin|flomax", case=False, na=False)
]["primaryid"].unique()
pv_df = demo[demo["primaryid"].isin(tamsulosin_case)].copy()
pv_df = pv_df[pv_df["sex"].isin(["M", "F"])]
pv_df = pd.merge(pv_df, reac, on="primaryid", how="inner")

top_reactions = pv_df["pt"].value_counts().head(20).index
results_list = []

total_female = len(pv_df[pv_df["sex"] == "F"])
total_male = len(pv_df[pv_df["sex"] == "M"])

for rx in top_reactions:
  a = len(pv_df[(pv_df["sex"] == "F") & (pv_df["pt"] == rx)])
  b = len(pv_df[(pv_df["sex"] == "M") & (pv_df["pt"] == rx)])
  c = total_female - a
  d = total_male - b

  if a == 0 or b == 0 or c == 0 or d == 0:
    continue

  ror = (a * d) / (b * c)
  se_ln_ror = np.sqrt((1 / a) + (1 / b) + (1 / c) + (1 / d))
  ci_lower = np.exp(np.log(ror) - 1.96 * se_ln_ror)
  ci_upper = np.exp(np.log(ror) + 1.96 * se_ln_ror)

  _, p_value, _, _ = stats.chi2_contingency([[a, b], [c, d]])

  results_list.append({
      "Reaction": rx,
      "ROR": ror,
      "CI_Lower": ci_lower,
      "CI_Upper": ci_upper,
      "P_Value": p_value,
  })

results_df = pd.DataFrame(results_list)

if not results_df.empty:
  _, p_adjusted, _, _ = multipletests(
      results_df["P_Value"], alpha=0.05, method="fdr_bh"
  )
  results_df["P_Adjusted"] = p_adjusted
  results_df["Signal"] = np.where(
      (results_df["CI_Lower"] > 1) & (results_df["P_Adjusted"] < 0.05),
      "YES (Significant)",
      "NO",
  )

results_df.to_csv("Tamsulosin_gender_signal_report.csv", index=False)
print("Pharmacovigilance analysis completed successfully!")
display(results_df)
```
